```python
import numpy as np
import matplotlib.pyplot as plt

# Real binding energy per nucleon data for some naturally occurring nuclei
# Data: (A, Binding energy per nucleon in MeV, Label)
nuclei_data = [
    (1, 0, 'H-1'),        # Hydrogen-1
    (2, 1.1, 'D-2'),      # Deuterium-2
    (3, 2.83, 'He-3'),    # Helium-3
    (4, 7.07, 'He-4'),    # Helium-4
    (6, 5.33, 'Li-6'),    # Lithium-6
    (7, 5.60, 'Li-7'),    # Lithium-7
    (12, 7.68, 'C-12'),   # Carbon-12
    (16, 7.98, 'O-16'),   # Oxygen-16
    (20, 8.45, 'Ne-20'),  # Neon-20
    (24, 8.53, 'Mg-24'),  # Magnesium-24
    (28, 8.79, 'Si-28'),  # Silicon-28
    (40, 8.55, 'Ca-40'),  # Calcium-40
    (56, 8.79, 'Fe-56'),  # Iron-56 (most stable nucleus)
    (63, 8.75, 'Cu-63'),  # Copper-63
    (197, 7.87, 'Au-197'),# Gold-197
    (238, 7.57, 'U-238')  # Uranium-238
]


# Convert data to arrays for plotting
A_values = np.array([item[0] for item in nuclei_data])
binding_energies = np.array([item[1] for item in nuclei_data])
labels = [item[2] for item in nuclei_data]

# Plotting the real data with labels
plt.figure(figsize=(10, 6))
#plt.plot(A_values, binding_energies, 'o-', label='Experimental Binding Energy per Nucleon', color='b', lw=2, markersize=6)
plt.plot(A_values, binding_energies, color='lightblue', lw=2, label='Binding Energy per Nucleon')  # Light blue line
plt.scatter(A_values, binding_energies, color='red', s=50, zorder=5)  # Red dots

# Adding labels to each data point
for i, label in enumerate(labels):
    if 'Ne' in label:
        continue
    if i < 3:
        plt.text(A_values[i] + 9.4, binding_energies[i] , label, fontsize=9, ha='center')
    elif i >= 4 and i <6:
         plt.text(A_values[i] + 9.4, binding_energies[i] , label, fontsize=9, ha='center')
    elif label == "Cu-63":
        plt.text(A_values[i]+2, binding_energies[i] + 0.2, label, fontsize=9, ha='center')
    elif label=="Ca-40":
        plt.text(A_values[i], binding_energies[i] + 0.2, label, fontsize=9, ha='center')
    else:
        plt.text(A_values[i]-5, binding_energies[i] + 0.2, label, fontsize=9, ha='center')

plt.title("Nuclear Binding Energy per Nucleon for Naturally Occurring Nuclei")
plt.xlabel("Mass Number A")
plt.ylabel("Binding Energy per Nucleon (MeV)")
plt.ylim([-0.5,9.5])
plt.grid(True)
plt.legend()
plt.show()

```
