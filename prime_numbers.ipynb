### Patterns of prime numbers plotted in polar coordinates 

A cool behaviour of prime numbers if plotted in polar coordinates where $r$ and $\theta$ are set as the same prime number

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sieve of Eratosthenes algorithm
# ref: https://en.wikipedia.org/wiki/Sieve_of_Eratosthenes
def get_prime_numbers_list(n):
    # prime numbers
    primes = []
    # list of numbers
    numbers = list(range(2, n+1))
    while numbers:
        # Assume the first number is prime
        prime = numbers.pop(0)
        # Add prime numbers to the list
        primes.append(prime)
        # remove all numbers divisible by the prime number
        numbers = [ x for x in numbers if x % prime != 0 ]

    # insert one
    primes.insert(0, 1)
    return primes

def from_polar_to_cartesian(numbers):
    r = numbers
    theta = numbers
    x = r * np.cos(numbers)
    y = r * np.sin(numbers)
    return x, y
```

```python
N = 20_000
primes = get_prime_numbers_list(N)
x, y = from_polar_to_cartesian(primes)
```

```python
plt.figure(figsize=(7,7))
plt.scatter(x, y, color="hotpink", s=10, alpha=0.5)
plt.title(f"First {N} prime numbers")
```

```python
N = 100_000
primes = get_prime_numbers_list(N)
x, y = from_polar_to_cartesian(primes)
```

```python
plt.figure(figsize=(7,7))
plt.scatter(x, y, color="blue", s=1, alpha=0.5)
plt.title(f"First {N} prime numbers")
```
